## Decision

Timeline posts entered the NLP file only when Twitter `LANG = en`. Keyword tweets have no `LANG` column; they were kept only when a **rule-based Latin-script / ASCII-ratio classifier** marked them `likely_english` after IFB tokens were stripped.

**English-ready is an operational inclusion rule, not proof that every retained post is native English.** Portuguese, Spanish, Indonesian, and Italian posts can survive because they are Latin-script. The Portuguese example below is a real retained post, kept on purpose.

```python
ex = pd.read_csv(OUT / "translation_examples.csv", encoding="utf-8-sig")
print("rows", len(ex))
print("columns", list(ex.columns))
display(ex)
# The Portuguese row is a real retained post, not a translation error.
pt = ex[ex.apply(lambda r: "dormir" in str(r.get("original_text", "")).lower(), axis=1)]
print("portuguese_rows_in_examples", len(pt))
if len(pt):
    print(pt.iloc[0].get("original_text") or pt.iloc[0].get("original"))
    print("status", pt.iloc[0].get("translation_status") or pt.iloc[0].get("status"))
```
